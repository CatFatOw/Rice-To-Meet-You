# Time-Series Forecasting of `AVERAGE_DAILY_VISITS`

## Forecast Period

**January 1, 2025 through December 31, 2025**

---

## Objective

Build a time-series model that predicts `AVERAGE_DAILY_VISITS` for every modeled `PLACEKEY` and every date from **2025-01-01** through **2025-12-31**.

Treat **2024-12-31** as the production forecast origin. Final predictions must use only information that was available on or before this date.

> **Leakage rule:** No visitor, spend, weather, or other feature recorded after `2024-12-31` may be used to generate the 2025 predictions.

---

## Input Data

**Parquet file:**

`/content/drive/MyDrive/RiceHackathon/StoreVisitCorePOIJoined/store_visits_core_poi_spend_12m.parquet`

The table contains approximately **7 million daily POI rows** covering **2020 through 2024**.

---

## Target Variable

`AVERAGE_DAILY_VISITS`

This is the value that the model will predict for each future POI-date row.

---

## Primary Row Identifier

Each row should be uniquely identified by:

- `PLACEKEY`
- `LOCAL_DATE`

The expected table grain is:

`PLACEKEY + LOCAL_DATE`

Before training, verify that this combination is unique and investigate any duplicate rows.

---

## Production Forecast Configuration

| Parameter | Value |
|---|---|
| Forecast origin | `2024-12-31` |
| First target date | `2025-01-01` |
| Last target date | `2025-12-31` |
| Maximum horizon | 365 days |
| Prediction target | `AVERAGE_DAILY_VISITS` |
| Location identifier | `PLACEKEY` |

## Step 1 — Verify grain uniqueness (PLACEKEY + LOCAL_DATE)

Before any feature engineering, confirm every `(PLACEKEY, LOCAL_DATE)` pair appears exactly once — duplicates would double-count a store-day and quietly corrupt every lag/rolling feature built on top of it.

```python
dupe_mask = joined_spend_df.duplicated(subset=["PLACEKEY", "LOCAL_DATE"], keep=False)
n_dupes = dupe_mask.sum()
print(f"Duplicate PLACEKEY + LOCAL_DATE rows: {n_dupes:,} ({n_dupes / len(joined_spend_df):.4%} of rows)")

dupe_placekeys = []
if n_dupes:
    # Select only the diagnostic columns, not the full 60+ column table --
    # pandas' pyarrow-backed string arrays hit a 32-bit offset overflow when
    # .loc[] row-subsets large text/geometry columns (e.g. POLYGON_WKT,
    # OPEN_HOURS). That overflow hits ANY row filter on the wide table, not
    # just display(), so the actual drop happens in Step 2 on a narrow
    # column projection instead of filtering joined_spend_df here.
    dupe_placekeys = joined_spend_df.loc[dupe_mask, "PLACEKEY"].unique().tolist()
    print(f"{len(dupe_placekeys)} PLACEKEYs are duplicated:")
    display(
        joined_spend_df.loc[dupe_mask, ["PLACEKEY", "LOCAL_DATE", "AVERAGE_DAILY_VISITS", "BRAND", "STORE_ID"]]
        .sort_values(["PLACEKEY", "LOCAL_DATE"])
        .head(20)
    )
    # These are auto-mall-style locations where one PLACEKEY/STORE_ID houses
    # multiple distinct brands (e.g. Honda + MINI), each with a genuinely
    # different AVERAGE_DAILY_VISITS reading -- not corrupted data, just
    # multiple businesses sharing one location identifier. Dropped rather
    # than summed, since blending unrelated brands' visit patterns into one
    # series would be a worse modeling assumption than losing 4 PLACEKEYs.
    print(f"Will drop these {len(dupe_placekeys)} PLACEKEYs in Step 2, once the table is narrowed to modeling columns.")
```

## Step 2 — Feature engineering (calendar + lag + rolling, past data only)

Same pattern as `add_features()` in `visitor/classical_ts_daily.py`, grouped by `PLACEKEY` instead of market. Every lag/rolling value is shifted from *strictly earlier* rows only — the leakage rule above means nothing computed here may look past a row's own `LOCAL_DATE`.

```python
import numpy as np
import lightgbm as lgb
from pandas.tseries.holiday import USFederalHolidayCalendar

TARGET = "AVERAGE_DAILY_VISITS"
FORECAST_ORIGIN = pd.Timestamp("2024-12-31")
TRAIN_END_YEAR = 2023      # train: 2020-2023
HOLDOUT_YEAR = 2024        # holdout: 2024 (mirrors classical_ts_daily.py's TEST_YEAR)
LAGS = (1, 7, 28, 365)
ROLLING_WINDOWS = (7, 28)

BASE_CATEGORICAL_COLUMNS = ["PLACEKEY"]
BASE_NUMERIC_COLUMNS = [
    "month", "year", "day_of_week", "day_of_month", "is_weekend", "is_federal_holiday",
    *[f"visits_lag_{lag}" for lag in LAGS],
    *[f"visits_rolling_{w}" for w in ROLLING_WINDOWS],
]
# CATEGORICAL_COLUMNS / NUMERIC_COLUMNS / FEATURE_COLUMNS (used by Steps 3-6)
# are finalized in Step 2.5 below, once ablation testing decides which extra
# feature groups get added on top of this base set.


def smape(actual: np.ndarray, predicted: np.ndarray) -> float:
    denom = np.abs(actual) + np.abs(predicted)
    ratio = np.where(denom == 0, 0.0, 2 * np.abs(actual - predicted) / np.where(denom == 0, 1, denom))
    return float(np.mean(ratio)) * 100


def score(actual: np.ndarray, predicted: np.ndarray) -> dict:
    predicted = np.clip(predicted, 0, None)  # visits can't be negative
    resid = actual - predicted
    ss_res = float(np.sum(resid ** 2))
    ss_tot = float(np.sum((actual - actual.mean()) ** 2))
    return {
        "MAE": float(np.mean(np.abs(resid))),
        "RMSE": float(np.sqrt(np.mean(resid ** 2))),
        "sMAPE": smape(actual, predicted),
        "R2": 1 - ss_res / ss_tot,
    }


def add_features(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df["LOCAL_DATE"] = pd.to_datetime(df["LOCAL_DATE"])
    df = df.sort_values(["PLACEKEY", "LOCAL_DATE"]).reset_index(drop=True)
    df["month"] = df["LOCAL_DATE"].dt.month
    df["year"] = df["LOCAL_DATE"].dt.year
    df["day_of_week"] = df["LOCAL_DATE"].dt.dayofweek
    df["day_of_month"] = df["LOCAL_DATE"].dt.day
    df["is_weekend"] = df["day_of_week"] >= 5
    holidays = USFederalHolidayCalendar().holidays(
        start=df["LOCAL_DATE"].min(), end=df["LOCAL_DATE"].max()
    )
    df["is_federal_holiday"] = df["LOCAL_DATE"].isin(holidays)

    visits = df.groupby("PLACEKEY")[TARGET]
    for lag in LAGS:
        df[f"visits_lag_{lag}"] = visits.shift(lag)
    for window in ROLLING_WINDOWS:
        df[f"visits_rolling_{window}"] = visits.transform(
            lambda s, size=window: s.shift(1).rolling(size, min_periods=1).mean()
        )
    return df


# Column-select down to the 3 modeling columns FIRST, then row-filter --
# doing it in this order (rather than filtering joined_spend_df directly)
# avoids the pyarrow offset overflow from Step 1: a plain column selection
# never touches row-level Arrow arrays, so it's safe regardless of how big
# POLYGON_WKT/OPEN_HOURS are, and PLACEKEY/LOCAL_DATE/TARGET are small
# enough that row-filtering this narrow frame is safe too.
narrow = joined_spend_df[["PLACEKEY", "LOCAL_DATE", TARGET]]
if dupe_placekeys:
    narrow = narrow.loc[~narrow["PLACEKEY"].isin(dupe_placekeys)]

featurized = add_features(narrow)
# LightGBM's pandas dtype check rejects pyarrow-backed extension dtypes
# (inherited from Step 0's dtype_backend="pyarrow" parquet read) --
# RandomForest/HistGB/XGBoost tolerated it, LightGBM does not. Cast just
# this narrow modeling frame to plain numpy float64, not the full 64-col
# table, so Step 0's memory-efficient load is untouched.
featurized[BASE_NUMERIC_COLUMNS + [TARGET]] = featurized[BASE_NUMERIC_COLUMNS + [TARGET]].astype("float64")
assert not featurized.duplicated(subset=["PLACEKEY", "LOCAL_DATE"]).any()
print(f"Featurized {len(featurized):,} rows")
print(f"\n{TARGET} skew: {featurized[TARGET].skew():.2f}  |  zero share: {(featurized[TARGET] == 0).mean():.2%}")
print("(classical_ts_daily.py's visitor_count used log1p because of skew ~4.9 -- re-check this number on real data before trusting the log1p below)")
```

## Step 2.5 — Feature-group ablation testing

Testing additional candidate feature groups on top of Step 2 features: business identity, geography, business type, physical POI, operating status, location structure, historical spending, plus richer rolling stats, calendar features, and spatial/neighborhood features. Rather than adding all of them speculatively, each group is added **one at a time** on top of the baseline and scored against a validation year; a group is kept only if it produces a **≥1% relative MAE reduction**

**This uses its own train/validation split, deliberately not the final 2024 holdout:**

| Split | Years | Used for |
|---|---|---|
| Ablation train | 2020–2022 | fit each candidate model |
| Ablation validation | 2023 | accept/reject each feature group |
| Final holdout | 2024 | checked exactly **once**, after group selection is locked in (Step 4) |

Re-using the 2024 holdout to make ~10 separate accept/reject decisions would let noise specific to that one year get "discovered" as a real pattern (repeated-testing / peeking bias). Carving 2023 out of train for this purpose keeps 2024 an honest, unpeeked check at the end — comparable to the `423.502041` LightGBM baseline Step 4 originally produced with the pre-ablation feature set.

**Ruled out during data profiling, before writing any feature code:**
- `BRAND_COUNT` — 99.92% null across PLACEKEYs (3 of 3,754 have a value). Dropped from "business identity."
- `CITY` and `MARKET_CORE` — identical to `MARKET_VISIT` for every PLACEKEY (checked directly, 100% match). Dropped from "geography" as pure duplicates.
- `CATEGORY_TAGS` — kept only as a tag-count number; `TOP_CATEGORY_2022`/`SUB_CATEGORY_2022` already carry most of the categorical signal, so a full multi-label expansion wasn't built.
- Spend leakage — `SPEND_TARGET_ELIGIBLE_DATE` was checked directly against `LOCAL_DATE`: 99.98% of spend-matched rows already satisfy `LOCAL_DATE >= SPEND_TARGET_ELIGIBLE_DATE`, meaning the source table's spend join is already as-of-safe. No extra lagging needed for "historical spending."
- `OPENED_ON` is only populated for 12.8% of PLACEKEYs — the "operating status" group's `days_since_opened` is NaN for the rest, capping how much that group can help.

```python
import json

# Fresh, non-Arrow-backed read for the extra candidate columns -- sidesteps
# the pyarrow ArrowDtype offset-overflow class of bugs from Step 1/2
# entirely (rather than trying to safely subset joined_spend_df's
# dtype_backend="pyarrow" columns), at the cost of one extra <2s parquet
# read of this 53MB file.
EXTRA_RAW_COLUMNS = [
    "PLACEKEY", "LOCAL_DATE",
    "BRAND", "MARKET_VISIT", "REGION", "LATITUDE", "LONGITUDE",
    "NAICS_CODE_2022", "SUB_CATEGORY_2022", "TOP_CATEGORY_2022", "CATEGORY_TAGS",
    "WKT_AREA_SQ_METERS", "POLYGON_CLASS", "ENCLOSED", "INCLUDES_PARKING_LOT",
    "OPENED_ON", "CLOSED_ON", "OPEN_HOURS", "PARENT_PLACEKEY", "IS_SYNTHETIC",
    "SPEND_MEDIAN_SPEND_PER_CUSTOMER", "SPEND_MEDIAN_SPEND_PER_TRANSACTION",
    "SPEND_ONLINE_SPEND", "SPEND_ONLINE_TRANSACTIONS",
    "SPEND_RAW_NUM_CUSTOMERS", "SPEND_RAW_NUM_TRANSACTIONS", "SPEND_RAW_TOTAL_SPEND",
    "SPEND_PCT_CHANGE_VS_PREV_MONTH", "SPEND_PCT_CHANGE_VS_PREV_YEAR",
    "SPEND_TARGET_ELIGIBLE_DATE",
]

_extra_raw = pd.read_parquet(JOINED_SPEND_FILE, columns=EXTRA_RAW_COLUMNS)
_extra_raw["LOCAL_DATE"] = pd.to_datetime(_extra_raw["LOCAL_DATE"])
if dupe_placekeys:
    _extra_raw = _extra_raw.loc[~_extra_raw["PLACEKEY"].isin(dupe_placekeys)]

# Leakage check: does the table's own spend join already respect an as-of cutoff?
_spend_matched = _extra_raw[_extra_raw["SPEND_RAW_TOTAL_SPEND"].notna()]
_violation_rate = (_spend_matched["LOCAL_DATE"] < _spend_matched["SPEND_TARGET_ELIGIBLE_DATE"]).mean()
print(f"Spend leakage check: {_violation_rate:.4%} of matched rows have LOCAL_DATE < SPEND_TARGET_ELIGIBLE_DATE "
      f"(should be ~0 -- confirms the join is already as-of-safe)")

SPEND_NUMERIC_COLUMNS = [
    "SPEND_MEDIAN_SPEND_PER_CUSTOMER", "SPEND_MEDIAN_SPEND_PER_TRANSACTION",
    "SPEND_ONLINE_SPEND", "SPEND_ONLINE_TRANSACTIONS",
    "SPEND_RAW_NUM_CUSTOMERS", "SPEND_RAW_NUM_TRANSACTIONS", "SPEND_RAW_TOTAL_SPEND",
    "SPEND_PCT_CHANGE_VS_PREV_MONTH", "SPEND_PCT_CHANGE_VS_PREV_YEAR",
]
spend_by_key = _extra_raw[["PLACEKEY", "LOCAL_DATE"] + SPEND_NUMERIC_COLUMNS]
static = _extra_raw.drop_duplicates(subset="PLACEKEY").reset_index(drop=True)
print(f"Static per-PLACEKEY attributes: {len(static):,} rows (one per PLACEKEY)")
```

```python
def group_business_identity(df: pd.DataFrame, static: pd.DataFrame) -> tuple[list[str], list[str]]:
    df["BRAND"] = df["PLACEKEY"].map(static.set_index("PLACEKEY")["BRAND"])
    return [], ["BRAND"]
    # BRAND_COUNT omitted -- 99.92% null, see Step 2.5 markdown.


def group_geography(df: pd.DataFrame, static: pd.DataFrame) -> tuple[list[str], list[str]]:
    lut = static.set_index("PLACEKEY")
    df["MARKET_VISIT"] = df["PLACEKEY"].map(lut["MARKET_VISIT"])
    df["REGION"] = df["PLACEKEY"].map(lut["REGION"])
    df["LATITUDE"] = df["PLACEKEY"].map(lut["LATITUDE"]).astype(float)
    df["LONGITUDE"] = df["PLACEKEY"].map(lut["LONGITUDE"]).astype(float)
    return ["LATITUDE", "LONGITUDE"], ["MARKET_VISIT", "REGION"]
    # CITY / MARKET_CORE omitted -- identical to MARKET_VISIT for every PLACEKEY.


def group_business_type(df: pd.DataFrame, static: pd.DataFrame) -> tuple[list[str], list[str]]:
    lut = static.set_index("PLACEKEY")
    df["NAICS_CODE_2022"] = df["PLACEKEY"].map(lut["NAICS_CODE_2022"]).astype("Int64").astype(str)
    df["SUB_CATEGORY_2022"] = df["PLACEKEY"].map(lut["SUB_CATEGORY_2022"])
    df["TOP_CATEGORY_2022"] = df["PLACEKEY"].map(lut["TOP_CATEGORY_2022"])
    tag_count = static["CATEGORY_TAGS"].apply(lambda s: len(json.loads(s)) if isinstance(s, str) else 0)
    df["category_tag_count"] = df["PLACEKEY"].map(dict(zip(static["PLACEKEY"], tag_count))).astype(float)
    return ["category_tag_count"], ["NAICS_CODE_2022", "SUB_CATEGORY_2022", "TOP_CATEGORY_2022"]
    # CATEGORY_TAGS itself kept as a tag-count number rather than multi-hot expansion --
    # TOP/SUB_CATEGORY_2022 already carry most of that signal.


def group_physical_poi(df: pd.DataFrame, static: pd.DataFrame) -> tuple[list[str], list[str]]:
    lut = static.set_index("PLACEKEY")
    df["WKT_AREA_SQ_METERS"] = df["PLACEKEY"].map(lut["WKT_AREA_SQ_METERS"]).astype(float)
    df["POLYGON_CLASS"] = df["PLACEKEY"].map(lut["POLYGON_CLASS"])
    df["ENCLOSED"] = df["PLACEKEY"].map(lut["ENCLOSED"]).astype(float)
    df["INCLUDES_PARKING_LOT"] = df["PLACEKEY"].map(lut["INCLUDES_PARKING_LOT"]).astype(float)
    return ["WKT_AREA_SQ_METERS", "ENCLOSED", "INCLUDES_PARKING_LOT"], ["POLYGON_CLASS"]


def _parse_open_hours_per_week(raw):
    if not isinstance(raw, str):
        return np.nan
    try:
        spec = json.loads(raw)
    except (ValueError, TypeError):
        return np.nan
    total_minutes = 0.0
    for ranges in spec.values():
        for start, end in ranges:
            sh, sm = map(int, start.split(":"))
            eh, em = map(int, end.split(":"))
            total_minutes += max(0, (eh * 60 + em) - (sh * 60 + sm))
    return total_minutes / 60.0


_DOW_KEYS = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]


def _parse_open_today_lookup(raw):
    if not isinstance(raw, str):
        return {i: np.nan for i in range(7)}
    try:
        spec = json.loads(raw)
    except (ValueError, TypeError):
        return {i: np.nan for i in range(7)}
    return {i: float(len(spec.get(k, [])) > 0) for i, k in enumerate(_DOW_KEYS)}


def group_operating_status(df: pd.DataFrame, static: pd.DataFrame) -> tuple[list[str], list[str]]:
    lut = static.set_index("PLACEKEY")
    opened = pd.to_datetime(df["PLACEKEY"].map(lut["OPENED_ON"]))
    local_date = pd.to_datetime(df["LOCAL_DATE"].astype("datetime64[ns]"))
    df["days_since_opened"] = (local_date - opened).dt.days.astype(float)
    df["has_closed_date"] = df["PLACEKEY"].map(lut["CLOSED_ON"].notna()).astype(float)
    hours_per_week = dict(zip(static["PLACEKEY"], static["OPEN_HOURS"].apply(_parse_open_hours_per_week)))
    df["open_hours_per_week"] = df["PLACEKEY"].map(hours_per_week)
    open_today_lut = dict(zip(static["PLACEKEY"], static["OPEN_HOURS"].apply(_parse_open_today_lookup)))
    df["is_open_today"] = [open_today_lut[pk].get(dow, np.nan) for pk, dow in zip(df["PLACEKEY"], df["day_of_week"])]
    return ["days_since_opened", "has_closed_date", "open_hours_per_week", "is_open_today"], []
    # days_since_opened is NaN for ~87% of PLACEKEYs (OPENED_ON mostly unrecorded).


def group_location_structure(df: pd.DataFrame, static: pd.DataFrame) -> tuple[list[str], list[str]]:
    lut = static.set_index("PLACEKEY")
    df["has_parent_placekey"] = df["PLACEKEY"].map(lut["PARENT_PLACEKEY"].notna()).astype(float)
    df["IS_SYNTHETIC"] = df["PLACEKEY"].map(lut["IS_SYNTHETIC"]).astype(float)
    return ["has_parent_placekey", "IS_SYNTHETIC"], []


def group_historical_spending(df: pd.DataFrame, spend_by_key: pd.DataFrame) -> tuple[list[str], list[str]]:
    # featurized["LOCAL_DATE"] is pyarrow-backed (timestamp[ns][pyarrow], inherited
    # from Step 0's dtype_backend="pyarrow" load via add_features' pd.to_datetime);
    # spend_by_key's LOCAL_DATE is plain datetime64[ns] (this section's fresh,
    # non-Arrow parquet read). pandas' merge refuses to join keys across those two
    # dtypes -- cast the join key on this side only, right before merging.
    left = df[["PLACEKEY", "LOCAL_DATE"]].copy()
    left["LOCAL_DATE"] = left["LOCAL_DATE"].astype("datetime64[ns]")
    merged = left.merge(spend_by_key, on=["PLACEKEY", "LOCAL_DATE"], how="left")
    for c in SPEND_NUMERIC_COLUMNS:
        df[c] = merged[c].to_numpy()
    return SPEND_NUMERIC_COLUMNS, []
    # Already verified LOCAL_DATE >= SPEND_TARGET_ELIGIBLE_DATE above -- the
    # source table's join is already as-of-safe, no extra lagging needed.


def group_rolling_extended(df: pd.DataFrame, _static: pd.DataFrame) -> tuple[list[str], list[str]]:
    visits = df.groupby("PLACEKEY")[TARGET]
    df["visits_rolling_median_7"] = visits.transform(lambda s: s.shift(1).rolling(7, min_periods=1).median())
    df["visits_rolling_mean_90"] = visits.transform(lambda s: s.shift(1).rolling(90, min_periods=1).mean())
    df["visits_rolling_std_28"] = visits.transform(lambda s: s.shift(1).rolling(28, min_periods=2).std())
    return ["visits_rolling_median_7", "visits_rolling_mean_90", "visits_rolling_std_28"], []


NAMED_HOLIDAY_RULES = {
    "is_new_years_day": lambda d: (d.dt.month == 1) & (d.dt.day == 1),
    "is_july_4th": lambda d: (d.dt.month == 7) & (d.dt.day == 4),
    "is_christmas_eve": lambda d: (d.dt.month == 12) & (d.dt.day == 24),
    "is_christmas_day": lambda d: (d.dt.month == 12) & (d.dt.day == 25),
    "is_new_years_eve": lambda d: (d.dt.month == 12) & (d.dt.day == 31),
}
CALENDAR_EXTENDED_NUMERIC_COLUMNS = [
    "week_of_year", "month_sin", "month_cos", "day_of_year_sin", "day_of_year_cos",
    "days_until_next_holiday", "days_since_last_holiday",
    *NAMED_HOLIDAY_RULES.keys(), "is_thanksgiving", "is_black_friday",
]


def calendar_extended_features(dates) -> pd.DataFrame:
    """Pure function of the calendar date -- reused for both historical rows
    (Step 2.5) and future forecast dates (Step 6), so the two can't drift
    out of sync. Named holidays matter separately from is_federal_holiday
    (Step 2) because retail visit patterns around e.g. Black Friday or
    Christmas Eve look nothing like a generic federal holiday such as
    Columbus Day."""
    d = pd.Series(pd.to_datetime(pd.Series(dates).astype("datetime64[ns]").to_numpy())).reset_index(drop=True)
    doy = d.dt.dayofyear
    out = pd.DataFrame(index=d.index)
    out["week_of_year"] = d.dt.isocalendar().week.astype(int).to_numpy()
    out["month_sin"] = np.sin(2 * np.pi * d.dt.month / 12)
    out["month_cos"] = np.cos(2 * np.pi * d.dt.month / 12)
    out["day_of_year_sin"] = np.sin(2 * np.pi * doy / 365.25)
    out["day_of_year_cos"] = np.cos(2 * np.pi * doy / 365.25)

    holidays = pd.DatetimeIndex(sorted(USFederalHolidayCalendar().holidays(
        start=d.min() - pd.Timedelta(days=30), end=d.max() + pd.Timedelta(days=30)
    )))
    hol_arr = holidays.values.astype("datetime64[D]")
    d_arr = d.values.astype("datetime64[D]")
    idx_after = np.searchsorted(hol_arr, d_arr, side="left")
    days_until = np.full(len(d_arr), np.nan)
    days_since = np.full(len(d_arr), np.nan)
    valid_after = idx_after < len(hol_arr)
    days_until[valid_after] = (hol_arr[idx_after[valid_after]] - d_arr[valid_after]).astype("timedelta64[D]").astype(float)
    valid_before = idx_after > 0
    days_since[valid_before] = (d_arr[valid_before] - hol_arr[idx_after[valid_before] - 1]).astype("timedelta64[D]").astype(float)
    out["days_until_next_holiday"] = days_until
    out["days_since_last_holiday"] = days_since

    for name, fn in NAMED_HOLIDAY_RULES.items():
        out[name] = fn(d).to_numpy().astype(float)

    # Thanksgiving is the 4th Thursday of November -- not a fixed calendar date.
    thursdays_nov = d[(d.dt.month == 11) & (d.dt.dayofweek == 3)]
    thanksgiving_dates = set(
        thursdays_nov.groupby(d.dt.year).apply(lambda s: s.sort_values().iloc[3] if len(s) >= 4 else pd.NaT)
    )
    thanksgiving_dates.discard(pd.NaT)
    black_friday_dates = {t + pd.Timedelta(days=1) for t in thanksgiving_dates}
    out["is_thanksgiving"] = d.isin(thanksgiving_dates).to_numpy().astype(float)
    out["is_black_friday"] = d.isin(black_friday_dates).to_numpy().astype(float)
    return out


def group_calendar_extended(df: pd.DataFrame, _static: pd.DataFrame) -> tuple[list[str], list[str]]:
    extra = calendar_extended_features(df["LOCAL_DATE"])
    for c in CALENDAR_EXTENDED_NUMERIC_COLUMNS:
        df[c] = extra[c].to_numpy()
    return CALENDAR_EXTENDED_NUMERIC_COLUMNS, []


def haversine_km(lat1, lon1, lat2, lon2):
    r = 6371.0
    p1, p2 = np.radians(lat1), np.radians(lat2)
    dphi = np.radians(lat2 - lat1)
    dlmb = np.radians(lon2 - lon1)
    a = np.sin(dphi / 2) ** 2 + np.cos(p1) * np.cos(p2) * np.sin(dlmb / 2) ** 2
    return 2 * r * np.arcsin(np.sqrt(a))


def group_spatial(df: pd.DataFrame, static: pd.DataFrame) -> tuple[list[str], list[str]]:
    s = static[["PLACEKEY", "MARKET_VISIT", "LATITUDE", "LONGITUDE"]].dropna(subset=["LATITUDE", "LONGITUDE"]).reset_index(drop=True)
    lat, lon = s["LATITUDE"].to_numpy(dtype=float), s["LONGITUDE"].to_numpy(dtype=float)
    nearby_count = np.zeros(len(s))
    dist_to_center = np.zeros(len(s))
    for _, idx in s.groupby("MARKET_VISIT").groups.items():
        idx = np.array(idx)
        mlat, mlon = lat[idx], lon[idx]
        center_lat, center_lon = mlat.mean(), mlon.mean()
        dist_to_center[idx] = haversine_km(mlat, mlon, center_lat, center_lon)
        for i in idx:
            nearby_count[i] = (haversine_km(lat[i], lon[i], mlat, mlon) <= 5.0).sum() - 1  # exclude self
    s["nearby_poi_count_5km"] = nearby_count
    s["distance_to_market_center_km"] = dist_to_center
    lut = s.set_index("PLACEKEY")
    df["nearby_poi_count_5km"] = df["PLACEKEY"].map(lut["nearby_poi_count_5km"])
    df["distance_to_market_center_km"] = df["PLACEKEY"].map(lut["distance_to_market_center_km"])
    return ["nearby_poi_count_5km", "distance_to_market_center_km"], []


FEATURE_GROUPS = {
    "business_identity": group_business_identity,
    "geography": group_geography,
    "business_type": group_business_type,
    "physical_poi": group_physical_poi,
    "operating_status": group_operating_status,
    "location_structure": group_location_structure,
    "historical_spending": lambda df, _s: group_historical_spending(df, spend_by_key),
    "rolling_extended": group_rolling_extended,
    "calendar_extended": group_calendar_extended,
    "spatial_neighborhood": group_spatial,
}
```

```python
ABLATION_TRAIN_END_YEAR = 2022   # ablation train: 2020-2022 (distinct from the final TRAIN_END_YEAR=2023 above)
ABLATION_VAL_YEAR = 2023         # ablation decisions made here -- the 2024 FINAL holdout stays untouched until Step 4
ACCEPT_THRESHOLD = 0.01          # >=1% relative MAE reduction on ABLATION_VAL_YEAR to keep a group


def _fit_predict_mae(train_df: pd.DataFrame, eval_df: pd.DataFrame, numeric_cols: list[str], categorical_cols: list[str]) -> float:
    feature_columns = categorical_cols + numeric_cols
    X_train = train_df[feature_columns].copy()
    X_eval = eval_df[feature_columns].copy()
    for col in categorical_cols:
        cats = pd.Categorical(X_train[col]).categories
        X_train[col] = pd.Categorical(X_train[col], categories=cats)
        X_eval[col] = pd.Categorical(X_eval[col], categories=cats)
    y_train = np.log1p(train_df[TARGET].to_numpy())
    model = lgb.LGBMRegressor(n_estimators=500, learning_rate=0.05, num_leaves=63, random_state=0, verbosity=-1)
    model.fit(X_train, y_train)
    pred = np.expm1(model.predict(X_eval))
    return score(eval_df[TARGET].to_numpy(), pred)["MAE"]


ablation_train_mask = featurized["year"] <= ABLATION_TRAIN_END_YEAR
ablation_val_mask = featurized["year"] == ABLATION_VAL_YEAR

base_train = featurized.loc[ablation_train_mask].dropna(subset=BASE_NUMERIC_COLUMNS + [TARGET])
base_val = featurized.loc[ablation_val_mask].dropna(subset=BASE_NUMERIC_COLUMNS + [TARGET])
baseline_val_mae = _fit_predict_mae(base_train, base_val, BASE_NUMERIC_COLUMNS, BASE_CATEGORICAL_COLUMNS)
print(f"Ablation baseline (train 2020-{ABLATION_TRAIN_END_YEAR}, val {ABLATION_VAL_YEAR}): MAE = {baseline_val_mae:.4f}")

group_columns: dict[str, tuple[list[str], list[str]]] = {}
ablation_rows = [{"group": "baseline", "val_MAE": baseline_val_mae, "delta_pct": 0.0, "accepted": None}]
accepted_groups: list[str] = []

for group_name, builder in FEATURE_GROUPS.items():
    added_numeric, added_categorical = builder(featurized, static)
    group_columns[group_name] = (added_numeric, added_categorical)
    train_g = featurized.loc[ablation_train_mask].dropna(subset=BASE_NUMERIC_COLUMNS + [TARGET])
    val_g = featurized.loc[ablation_val_mask].dropna(subset=BASE_NUMERIC_COLUMNS + [TARGET])
    mae = _fit_predict_mae(
        train_g, val_g,
        BASE_NUMERIC_COLUMNS + added_numeric, BASE_CATEGORICAL_COLUMNS + added_categorical,
    )
    delta_pct = (baseline_val_mae - mae) / baseline_val_mae * 100
    accepted = delta_pct >= ACCEPT_THRESHOLD * 100
    print(f"{group_name}: val_MAE={mae:.4f}  delta={delta_pct:+.2f}%  -> {'ACCEPT' if accepted else 'reject'}")
    ablation_rows.append({"group": group_name, "val_MAE": mae, "delta_pct": delta_pct, "accepted": accepted})
    if accepted:
        accepted_groups.append(group_name)

ablation_summary = pd.DataFrame(ablation_rows)
display(ablation_summary)
print(f"\nAccepted groups: {accepted_groups}")
```

```python
ACCEPTED_NUMERIC_COLUMNS = [c for g in accepted_groups for c in group_columns[g][0]]
ACCEPTED_CATEGORICAL_COLUMNS = [c for g in accepted_groups for c in group_columns[g][1]]

CATEGORICAL_COLUMNS = BASE_CATEGORICAL_COLUMNS + ACCEPTED_CATEGORICAL_COLUMNS
NUMERIC_COLUMNS = BASE_NUMERIC_COLUMNS + ACCEPTED_NUMERIC_COLUMNS
FEATURE_COLUMNS = CATEGORICAL_COLUMNS + NUMERIC_COLUMNS
featurized[ACCEPTED_NUMERIC_COLUMNS] = featurized[ACCEPTED_NUMERIC_COLUMNS].astype("float64")

print(f"Final feature set ({len(FEATURE_COLUMNS)} columns):")
print(FEATURE_COLUMNS)
```

## Step 3 — Chronological split + native-categorical frame

`PLACEKEY` cardinality makes one-hot encoding impractical for most of Step 4's candidates — pass it through as a pandas `category` dtype instead, the same `native_categorical_frame` pattern the retired `train.py` used for high-cardinality columns. Holdout categories are aligned to train's, so a PLACEKEY the model never saw in training becomes a missing code rather than a new split it has no learned rule for. (RandomForest and HistGradientBoosting skip this and use ordinal-encoded PLACEKEY instead — see Step 4; HistGradientBoosting's native categorical support caps at 255 categories, well under PLACEKEY's cardinality.)

```python
def chronological_split(df: pd.DataFrame) -> tuple[pd.DataFrame, pd.DataFrame]:
    train = df[df["year"] <= TRAIN_END_YEAR].copy()
    holdout = df[df["year"] == HOLDOUT_YEAR].copy()
    return train, holdout


def native_categorical_frame(
    train: pd.DataFrame, test: pd.DataFrame, feature_columns: list[str] = FEATURE_COLUMNS
) -> tuple[pd.DataFrame, pd.DataFrame]:
    X_train = train[feature_columns].copy()
    X_test = test[feature_columns].copy()
    for col in CATEGORICAL_COLUMNS:
        cats = pd.Categorical(X_train[col]).categories
        X_train[col] = pd.Categorical(X_train[col], categories=cats)
        X_test[col] = pd.Categorical(X_test[col], categories=cats)
        # A test-set PLACEKEY absent from train becomes NaN here (categories
        # restricted to train's, by design -- see the markdown above). XGBoost
        # and LightGBM treat that NaN as "missing" and predict fine. CatBoost
        # is the opposite: its Pool rejects NaN in a cat_feature column, but
        # errors just as hard on an unrecognized string category -- it needs
        # a per-model sentinel, applied only at its own predict call (see
        # Step 4), not shared here.
    return X_train, X_test


train_df, holdout_df = chronological_split(featurized)
# visits_lag_365 needs a full prior year of history -- rows without it (each
# PLACEKEY's first ~365 days) get dropped rather than imputed, same
# discipline classical_ts_daily.py/train_monthly.py used.
before = len(train_df)
train_df = train_df.dropna(subset=NUMERIC_COLUMNS + [TARGET])
holdout_df = holdout_df.dropna(subset=NUMERIC_COLUMNS + [TARGET])
print(f"Train: {len(train_df):,} rows (2020-{TRAIN_END_YEAR}), dropped {before - len(train_df):,} missing lag/rolling history")
print(f"Holdout: {len(holdout_df):,} rows ({HOLDOUT_YEAR})")
```

## Step 4 — Model bake-off vs. the lag-365 seasonal baseline, scored on the 2024 holdout

Same idea as `classical_ts_daily.py`'s bake-off: don't commit to one algorithm, score several against the naive baseline and let the holdout pick a winner. Candidates are the native-categorical models `train.py` used for its own high-cardinality columns (HistGradientBoosting, XGBoost, LightGBM, CatBoost) plus RandomForest via ordinal encoding. The lag-365 baseline ("this day = the same day one year ago") is still the real bar to clear — `classical_ts_daily.py`'s month-ahead bake-off found a naive baseline beating every trained model once forecasts turn recursive, so don't assume a trained model wins by default.

```python
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import HistGradientBoostingRegressor, RandomForestRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OrdinalEncoder
import xgboost as xgb
from catboost import CatBoostRegressor

# HistGradientBoosting is NOT in this set, unlike train.py's old bake-off --
# its native categorical support caps at max_bins (255 categories), which
# worked for train.py's "_grouped" columns (~190 brand groups, deliberately
# kept low-cardinality) but PLACEKEY has far more distinct values than that.
# HistGradientBoosting instead falls through to ordinal_pipeline below,
# same as RandomForest.
NATIVE_CATEGORICAL_MODELS = ("XGBoost", "LightGBM", "CatBoost")


def build_model(name: str):
    if name == "RandomForest":
        return RandomForestRegressor(n_estimators=300, max_depth=14, min_samples_leaf=5, n_jobs=-1, random_state=0)
    if name == "HistGradientBoosting":
        return HistGradientBoostingRegressor(max_iter=400, random_state=0)
    if name == "XGBoost":
        return xgb.XGBRegressor(
            n_estimators=400, max_depth=6, learning_rate=0.05, tree_method="hist",
            enable_categorical=True, random_state=0,
        )
    if name == "LightGBM":
        return lgb.LGBMRegressor(n_estimators=500, learning_rate=0.05, num_leaves=63, random_state=0, verbosity=-1)
    if name == "CatBoost":
        return CatBoostRegressor(
            iterations=400, depth=8, learning_rate=0.05, cat_features=CATEGORICAL_COLUMNS,
            random_state=0, verbose=False,
        )
    raise ValueError(f"no builder for {name!r}")


def ordinal_pipeline(estimator) -> Pipeline:
    """RandomForest and HistGradientBoosting don't get native categorical
    support here -- ordinal-encode PLACEKEY instead of one-hot (same
    shortcut train.py used for its own high-cardinality categoricals;
    one-hot over thousands of PLACEKEYs would blow up split search for no
    benefit)."""
    pre = ColumnTransformer([
        ("cat", OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1), CATEGORICAL_COLUMNS),
        ("num", "passthrough", NUMERIC_COLUMNS),
    ])
    return Pipeline([("pre", pre), ("model", estimator)])


def catboost_categorical_nan_guard(X: pd.DataFrame) -> pd.DataFrame:
    """CatBoost's Pool rejects NaN in a cat_feature column outright --
    unlike XGBoost/LightGBM, which both treat it as "missing" and split on
    it natively -- at TRAIN time as well as predict time. Two distinct
    sources of NaN hit this: SUB_CATEGORY_2022 is naturally null for 0.11%
    of PLACEKEYs even inside the training set, and a holdout/forecast
    PLACEKEY absent from train's categories (see native_categorical_frame)
    produces NaN too. Both get the same sentinel-category fix, applied to
    every categorical column CatBoost sees, not just PLACEKEY."""
    X = X.copy()
    for col in CATEGORICAL_COLUMNS:
        if X[col].isna().any():
            X[col] = X[col].cat.add_categories(["__unseen__"]).fillna("__unseen__")
    return X


X_train, X_holdout = native_categorical_frame(train_df, holdout_df)
y_train = np.log1p(train_df[TARGET].to_numpy())  # right-skewed visit counts -- re-verify against the skew printed in Step 2
y_holdout = holdout_df[TARGET].to_numpy()

rows = [{"model": "baseline_seasonal_lag365", **score(y_holdout, holdout_df["visits_lag_365"].to_numpy())}]

fitted = {}
for name in ("RandomForest", "HistGradientBoosting", "XGBoost", "LightGBM", "CatBoost"):
    print(f"Fitting {name}...")
    if name in NATIVE_CATEGORICAL_MODELS:
        model = build_model(name)
        train_X = catboost_categorical_nan_guard(X_train) if name == "CatBoost" else X_train
        model.fit(train_X, y_train)
        holdout_X = catboost_categorical_nan_guard(X_holdout) if name == "CatBoost" else X_holdout
        pred = np.expm1(model.predict(holdout_X))
    else:
        model = ordinal_pipeline(build_model(name))
        model.fit(train_df[FEATURE_COLUMNS], y_train)
        pred = np.expm1(model.predict(holdout_df[FEATURE_COLUMNS]))
    fitted[name] = model
    rows.append({"model": name, **score(y_holdout, pred)})

comparison = pd.DataFrame(rows).sort_values("MAE").reset_index(drop=True)
display(comparison)

winner_name = comparison.iloc[0]["model"]
print(f"\nBest model by MAE: {winner_name}")
```

## Step 5 — Refit the bake-off winner through 2024-12-31, generate one prediction per PLACEKEY per day for 2025

Rolls the winning model (or the lag-365 baseline itself, if nothing beat it in Step 4) forward one day at a time: each day's prediction feeds `visits_lag_1/7/28` for the next day, since there's no ground truth yet for a date that hasn't happened. `visits_lag_365` stays a *real* 2024 observation for the entire 2025 horizon, which is what keeps this from degrading the way a pure short-lag recursive rollout would — `classical_ts_daily.py`'s month-ahead bake-off already showed 31 days of that kind of compounding losing to a naive baseline, and 365 days on top of a short-lag model would likely be worse. If Step 4's winner didn't clearly beat the baseline, this cell is effectively just replaying the baseline forward.

PLACEKEYs with less than 365 days of history by the forecast origin are skipped (no real `visits_lag_365` to build from) rather than imputed.

```python
final_train = featurized.dropna(subset=NUMERIC_COLUMNS + [TARGET])
y_final = np.log1p(final_train[TARGET].to_numpy())

train_categories = None
if winner_name == "baseline_seasonal_lag365":
    final_model = None
elif winner_name in NATIVE_CATEGORICAL_MODELS:
    X_final, _ = native_categorical_frame(final_train, final_train)
    final_model = build_model(winner_name)
    fit_X_final = catboost_categorical_nan_guard(X_final) if winner_name == "CatBoost" else X_final
    final_model.fit(fit_X_final, y_final)
    train_categories = {col: X_final[col].cat.categories for col in CATEGORICAL_COLUMNS}
else:
    final_model = ordinal_pipeline(build_model(winner_name))
    final_model.fit(final_train[FEATURE_COLUMNS], y_final)

# Ablation-accepted groups (business_identity, business_type, calendar_extended)
# don't need recursion the way visit lags do: BRAND/NAICS_CODE_2022/etc. are
# fixed per PLACEKEY for the whole horizon, and the calendar_extended columns
# are a pure function of the forecast date -- both are looked up/computed
# once up front rather than carried through the day-by-day loop below.
STATIC_EXTRA_COLUMNS = [
    c for c in ACCEPTED_CATEGORICAL_COLUMNS + ACCEPTED_NUMERIC_COLUMNS
    if c not in CALENDAR_EXTENDED_NUMERIC_COLUMNS
]
static_extra_lookup = (
    featurized.drop_duplicates(subset="PLACEKEY").set_index("PLACEKEY")[STATIC_EXTRA_COLUMNS].to_dict(orient="index")
    if STATIC_EXTRA_COLUMNS else {}
)

forecast_dates_2025 = pd.date_range("2025-01-01", "2025-12-31", freq="D")
calendar_extended_by_date = (
    calendar_extended_features(pd.Series(forecast_dates_2025)).set_axis(forecast_dates_2025)
    if "calendar_extended" in accepted_groups else None
)


def recursive_placekey_forecast(history: pd.DataFrame, forecast_dates: pd.DatetimeIndex) -> pd.DataFrame:
    history = history.sort_values(["PLACEKEY", "LOCAL_DATE"])
    holidays = set(USFederalHolidayCalendar().holidays(start=forecast_dates.min(), end=forecast_dates.max()))

    series = {
        placekey: group[TARGET].astype(float).tolist()
        for placekey, group in history.groupby("PLACEKEY")
    }

    predictions = []
    for forecast_date in forecast_dates:
        calendar_row = calendar_extended_by_date.loc[forecast_date].to_dict() if calendar_extended_by_date is not None else {}
        rows = []
        placekeys = []
        for placekey, values in series.items():
            if len(values) < 365:
                continue  # no real visits_lag_365 yet for this store -- skip rather than impute
            rows.append({
                "PLACEKEY": placekey,
                "month": forecast_date.month,
                "year": forecast_date.year,
                "day_of_week": forecast_date.dayofweek,
                "day_of_month": forecast_date.day,
                "is_weekend": forecast_date.dayofweek >= 5,
                "is_federal_holiday": forecast_date in holidays,
                "visits_lag_1": values[-1],
                "visits_lag_7": values[-7],
                "visits_lag_28": values[-28],
                "visits_lag_365": values[-365],
                "visits_rolling_7": float(np.mean(values[-7:])),
                "visits_rolling_28": float(np.mean(values[-28:])),
                **static_extra_lookup.get(placekey, {}),
                **calendar_row,
            })
            placekeys.append(placekey)

        features = pd.DataFrame(rows)
        if winner_name == "baseline_seasonal_lag365":
            predicted = features["visits_lag_365"].to_numpy()
        else:
            if winner_name in NATIVE_CATEGORICAL_MODELS:
                for col in CATEGORICAL_COLUMNS:
                    features[col] = pd.Categorical(features[col], categories=train_categories[col])
                if winner_name == "CatBoost":
                    features = catboost_categorical_nan_guard(features)
            predicted = np.clip(np.expm1(final_model.predict(features[FEATURE_COLUMNS])), 0, None)

        for placekey, value in zip(placekeys, predicted):
            value = float(value)
            series[placekey].append(value)
            predictions.append({
                "PLACEKEY": placekey,
                "LOCAL_DATE": forecast_date,
                f"predicted_{TARGET}": value,
            })
    return pd.DataFrame(predictions)


history_through_origin = featurized[featurized["LOCAL_DATE"] <= FORECAST_ORIGIN]

# 365 sequential predict calls, one per PLACEKEY-day -- expect this cell to
# take a while on the full dataset.
predictions_2025 = recursive_placekey_forecast(history_through_origin, forecast_dates_2025)
print(f"Generated {len(predictions_2025):,} predictions across {predictions_2025['PLACEKEY'].nunique():,} PLACEKEYs")

# Absolute, matching JOINED_SPEND_FILE's style above -- a relative path
# here resolves against the KERNEL's cwd (this notebook's own directory
# when run via nbconvert), not the repo root, and silently wrote the
# first full run into visitor/visitor_data/... instead of visitor_data/...
OUTPUT_PATH = Path(
    "/Users/rusli/Documents/GitHub/Rice-To-Meet-You/"
    "visitor_data/output/visitor_predictions_2025.csv"
)
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)
predictions_2025.to_csv(OUTPUT_PATH, index=False)
print(f"Wrote -> {OUTPUT_PATH}")
```
