# Time-Series Forecasting of `AVERAGE_DAILY_VISITS`

## Forecast Period

**January 1, 2025 through December 31, 2025**

---

## Objective

Build a time-series model that predicts `AVERAGE_DAILY_VISITS` for every modeled `PLACEKEY` and every date from **2025-01-01** through **2025-12-31**.

Treat **2024-12-31** as the production forecast origin. Final predictions must use only information that was available on or before this date.

> **Leakage rule:** No visitor, spend, weather, or other feature recorded after `2024-12-31` may be used to generate the 2025 predictions.

---

## Input Data

**Parquet file:**

`/content/drive/MyDrive/RiceHackathon/StoreVisitCorePOIJoined/store_visits_core_poi_spend_12m.parquet`

The table contains approximately **7 million daily POI rows** covering **2020 through 2024**.

---

## Target Variable

`AVERAGE_DAILY_VISITS`

This is the value that the model will predict for each future POI-date row.

---

## Primary Row Identifier

Each row should be uniquely identified by:

- `PLACEKEY`
- `LOCAL_DATE`

The expected table grain is:

`PLACEKEY + LOCAL_DATE`

Before training, verify that this combination is unique and investigate any duplicate rows.

---

## Production Forecast Configuration

| Parameter | Value |
|---|---|
| Forecast origin | `2024-12-31` |
| First target date | `2025-01-01` |
| Last target date | `2025-12-31` |
| Maximum horizon | 365 days |
| Prediction target | `AVERAGE_DAILY_VISITS` |
| Location identifier | `PLACEKEY` |

## Step 1 — Verify grain uniqueness (PLACEKEY + LOCAL_DATE)

Before any feature engineering, confirm every `(PLACEKEY, LOCAL_DATE)` pair appears exactly once — duplicates would double-count a store-day and quietly corrupt every lag/rolling feature built on top of it.

```python
dupe_mask = joined_spend_df.duplicated(subset=["PLACEKEY", "LOCAL_DATE"], keep=False)
n_dupes = dupe_mask.sum()
print(f"Duplicate PLACEKEY + LOCAL_DATE rows: {n_dupes:,} ({n_dupes / len(joined_spend_df):.4%} of rows)")

dupe_placekeys = []
if n_dupes:
    # Select only the diagnostic columns, not the full 60+ column table --
    # pandas' pyarrow-backed string arrays hit a 32-bit offset overflow when
    # .loc[] row-subsets large text/geometry columns (e.g. POLYGON_WKT,
    # OPEN_HOURS). That overflow hits ANY row filter on the wide table, not
    # just display(), so the actual drop happens in Step 2 on a narrow
    # column projection instead of filtering joined_spend_df here.
    dupe_placekeys = joined_spend_df.loc[dupe_mask, "PLACEKEY"].unique().tolist()
    print(f"{len(dupe_placekeys)} PLACEKEYs are duplicated:")
    display(
        joined_spend_df.loc[dupe_mask, ["PLACEKEY", "LOCAL_DATE", "AVERAGE_DAILY_VISITS", "BRAND", "STORE_ID"]]
        .sort_values(["PLACEKEY", "LOCAL_DATE"])
        .head(20)
    )
    # These are auto-mall-style locations where one PLACEKEY/STORE_ID houses
    # multiple distinct brands (e.g. Honda + MINI), each with a genuinely
    # different AVERAGE_DAILY_VISITS reading -- not corrupted data, just
    # multiple businesses sharing one location identifier. Dropped rather
    # than summed, since blending unrelated brands' visit patterns into one
    # series would be a worse modeling assumption than losing 4 PLACEKEYs.
    print(f"Will drop these {len(dupe_placekeys)} PLACEKEYs in Step 2, once the table is narrowed to modeling columns.")
```

## Step 2 — Feature engineering (calendar + lag + rolling, past data only)

Same pattern as `add_features()` in `visitor/classical_ts_daily.py`, grouped by `PLACEKEY` instead of market. Every lag/rolling value is shifted from *strictly earlier* rows only — the leakage rule above means nothing computed here may look past a row's own `LOCAL_DATE`.

```python
import numpy as np
import lightgbm as lgb
from pandas.tseries.holiday import USFederalHolidayCalendar

TARGET = "AVERAGE_DAILY_VISITS"
FORECAST_ORIGIN = pd.Timestamp("2024-12-31")
TRAIN_END_YEAR = 2023      # train: 2020-2023
HOLDOUT_YEAR = 2024        # holdout: 2024 (mirrors classical_ts_daily.py's TEST_YEAR)
LAGS = (1, 7, 28, 365)
ROLLING_WINDOWS = (7, 28)

CATEGORICAL_COLUMNS = ["PLACEKEY"]
NUMERIC_COLUMNS = [
    "month", "year", "day_of_week", "day_of_month", "is_weekend", "is_federal_holiday",
    *[f"visits_lag_{lag}" for lag in LAGS],
    *[f"visits_rolling_{w}" for w in ROLLING_WINDOWS],
]
FEATURE_COLUMNS = CATEGORICAL_COLUMNS + NUMERIC_COLUMNS


def smape(actual: np.ndarray, predicted: np.ndarray) -> float:
    denom = np.abs(actual) + np.abs(predicted)
    ratio = np.where(denom == 0, 0.0, 2 * np.abs(actual - predicted) / np.where(denom == 0, 1, denom))
    return float(np.mean(ratio)) * 100


def score(actual: np.ndarray, predicted: np.ndarray) -> dict:
    predicted = np.clip(predicted, 0, None)  # visits can't be negative
    resid = actual - predicted
    ss_res = float(np.sum(resid ** 2))
    ss_tot = float(np.sum((actual - actual.mean()) ** 2))
    return {
        "MAE": float(np.mean(np.abs(resid))),
        "RMSE": float(np.sqrt(np.mean(resid ** 2))),
        "sMAPE": smape(actual, predicted),
        "R2": 1 - ss_res / ss_tot,
    }


def add_features(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df["LOCAL_DATE"] = pd.to_datetime(df["LOCAL_DATE"])
    df = df.sort_values(["PLACEKEY", "LOCAL_DATE"]).reset_index(drop=True)
    df["month"] = df["LOCAL_DATE"].dt.month
    df["year"] = df["LOCAL_DATE"].dt.year
    df["day_of_week"] = df["LOCAL_DATE"].dt.dayofweek
    df["day_of_month"] = df["LOCAL_DATE"].dt.day
    df["is_weekend"] = df["day_of_week"] >= 5
    holidays = USFederalHolidayCalendar().holidays(
        start=df["LOCAL_DATE"].min(), end=df["LOCAL_DATE"].max()
    )
    df["is_federal_holiday"] = df["LOCAL_DATE"].isin(holidays)

    visits = df.groupby("PLACEKEY")[TARGET]
    for lag in LAGS:
        df[f"visits_lag_{lag}"] = visits.shift(lag)
    for window in ROLLING_WINDOWS:
        df[f"visits_rolling_{window}"] = visits.transform(
            lambda s, size=window: s.shift(1).rolling(size, min_periods=1).mean()
        )
    return df


# Column-select down to the 3 modeling columns FIRST, then row-filter --
# doing it in this order (rather than filtering joined_spend_df directly)
# avoids the pyarrow offset overflow from Step 1: a plain column selection
# never touches row-level Arrow arrays, so it's safe regardless of how big
# POLYGON_WKT/OPEN_HOURS are, and PLACEKEY/LOCAL_DATE/TARGET are small
# enough that row-filtering this narrow frame is safe too.
narrow = joined_spend_df[["PLACEKEY", "LOCAL_DATE", TARGET]]
if dupe_placekeys:
    narrow = narrow.loc[~narrow["PLACEKEY"].isin(dupe_placekeys)]

featurized = add_features(narrow)
# LightGBM's pandas dtype check rejects pyarrow-backed extension dtypes
# (inherited from Step 0's dtype_backend="pyarrow" parquet read) --
# RandomForest/HistGB/XGBoost tolerated it, LightGBM does not. Cast just
# this narrow modeling frame to plain numpy float64, not the full 64-col
# table, so Step 0's memory-efficient load is untouched.
featurized[NUMERIC_COLUMNS + [TARGET]] = featurized[NUMERIC_COLUMNS + [TARGET]].astype("float64")
assert not featurized.duplicated(subset=["PLACEKEY", "LOCAL_DATE"]).any()
print(f"Featurized {len(featurized):,} rows")
print(f"\n{TARGET} skew: {featurized[TARGET].skew():.2f}  |  zero share: {(featurized[TARGET] == 0).mean():.2%}")
print("(classical_ts_daily.py's visitor_count used log1p because of skew ~4.9 -- re-check this number on real data before trusting the log1p below)")
```

## Step 3 — Chronological split + native-categorical frame

`PLACEKEY` cardinality makes one-hot encoding impractical for most of Step 4's candidates — pass it through as a pandas `category` dtype instead, the same `native_categorical_frame` pattern the retired `train.py` used for high-cardinality columns. Holdout categories are aligned to train's, so a PLACEKEY the model never saw in training becomes a missing code rather than a new split it has no learned rule for. (RandomForest and HistGradientBoosting skip this and use ordinal-encoded PLACEKEY instead — see Step 4; HistGradientBoosting's native categorical support caps at 255 categories, well under PLACEKEY's cardinality.)

```python
def chronological_split(df: pd.DataFrame) -> tuple[pd.DataFrame, pd.DataFrame]:
    train = df[df["year"] <= TRAIN_END_YEAR].copy()
    holdout = df[df["year"] == HOLDOUT_YEAR].copy()
    return train, holdout


def native_categorical_frame(
    train: pd.DataFrame, test: pd.DataFrame, feature_columns: list[str] = FEATURE_COLUMNS
) -> tuple[pd.DataFrame, pd.DataFrame]:
    X_train = train[feature_columns].copy()
    X_test = test[feature_columns].copy()
    for col in CATEGORICAL_COLUMNS:
        cats = pd.Categorical(X_train[col]).categories
        X_train[col] = pd.Categorical(X_train[col], categories=cats)
        X_test[col] = pd.Categorical(X_test[col], categories=cats)
        # A test-set PLACEKEY absent from train becomes NaN here (categories
        # restricted to train's, by design -- see the markdown above). XGBoost
        # and LightGBM treat that NaN as "missing" and predict fine. CatBoost
        # is the opposite: its Pool rejects NaN in a cat_feature column, but
        # errors just as hard on an unrecognized string category -- it needs
        # a per-model sentinel, applied only at its own predict call (see
        # Step 4), not shared here.
    return X_train, X_test


train_df, holdout_df = chronological_split(featurized)
# visits_lag_365 needs a full prior year of history -- rows without it (each
# PLACEKEY's first ~365 days) get dropped rather than imputed, same
# discipline classical_ts_daily.py/train_monthly.py used.
before = len(train_df)
train_df = train_df.dropna(subset=NUMERIC_COLUMNS + [TARGET])
holdout_df = holdout_df.dropna(subset=NUMERIC_COLUMNS + [TARGET])
print(f"Train: {len(train_df):,} rows (2020-{TRAIN_END_YEAR}), dropped {before - len(train_df):,} missing lag/rolling history")
print(f"Holdout: {len(holdout_df):,} rows ({HOLDOUT_YEAR})")
```

## Step 4 — Model bake-off vs. the lag-365 seasonal baseline, scored on the 2024 holdout

Same idea as `classical_ts_daily.py`'s bake-off: don't commit to one algorithm, score several against the naive baseline and let the holdout pick a winner. Candidates are the native-categorical models `train.py` used for its own high-cardinality columns (HistGradientBoosting, XGBoost, LightGBM, CatBoost) plus RandomForest via ordinal encoding. The lag-365 baseline ("this day = the same day one year ago") is still the real bar to clear — `classical_ts_daily.py`'s month-ahead bake-off found a naive baseline beating every trained model once forecasts turn recursive, so don't assume a trained model wins by default.

```python
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import HistGradientBoostingRegressor, RandomForestRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OrdinalEncoder
import xgboost as xgb
from catboost import CatBoostRegressor

# HistGradientBoosting is NOT in this set, unlike train.py's old bake-off --
# its native categorical support caps at max_bins (255 categories), which
# worked for train.py's "_grouped" columns (~190 brand groups, deliberately
# kept low-cardinality) but PLACEKEY has far more distinct values than that.
# HistGradientBoosting instead falls through to ordinal_pipeline below,
# same as RandomForest.
NATIVE_CATEGORICAL_MODELS = ("XGBoost", "LightGBM", "CatBoost")


def build_model(name: str):
    if name == "RandomForest":
        return RandomForestRegressor(n_estimators=300, max_depth=14, min_samples_leaf=5, n_jobs=-1, random_state=0)
    if name == "HistGradientBoosting":
        return HistGradientBoostingRegressor(max_iter=400, random_state=0)
    if name == "XGBoost":
        return xgb.XGBRegressor(
            n_estimators=400, max_depth=6, learning_rate=0.05, tree_method="hist",
            enable_categorical=True, random_state=0,
        )
    if name == "LightGBM":
        return lgb.LGBMRegressor(n_estimators=500, learning_rate=0.05, num_leaves=63, random_state=0, verbosity=-1)
    if name == "CatBoost":
        return CatBoostRegressor(
            iterations=400, depth=8, learning_rate=0.05, cat_features=CATEGORICAL_COLUMNS,
            random_state=0, verbose=False,
        )
    raise ValueError(f"no builder for {name!r}")


def ordinal_pipeline(estimator) -> Pipeline:
    """RandomForest and HistGradientBoosting don't get native categorical
    support here -- ordinal-encode PLACEKEY instead of one-hot (same
    shortcut train.py used for its own high-cardinality categoricals;
    one-hot over thousands of PLACEKEYs would blow up split search for no
    benefit)."""
    pre = ColumnTransformer([
        ("cat", OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1), CATEGORICAL_COLUMNS),
        ("num", "passthrough", NUMERIC_COLUMNS),
    ])
    return Pipeline([("pre", pre), ("model", estimator)])


X_train, X_holdout = native_categorical_frame(train_df, holdout_df)
y_train = np.log1p(train_df[TARGET].to_numpy())  # right-skewed visit counts -- re-verify against the skew printed in Step 2
y_holdout = holdout_df[TARGET].to_numpy()

rows = [{"model": "baseline_seasonal_lag365", **score(y_holdout, holdout_df["visits_lag_365"].to_numpy())}]

fitted = {}
for name in ("RandomForest", "HistGradientBoosting", "XGBoost", "LightGBM", "CatBoost"):
    print(f"Fitting {name}...")
    if name in NATIVE_CATEGORICAL_MODELS:
        model = build_model(name)
        model.fit(X_train, y_train)
        holdout_X = X_holdout
        if name == "CatBoost" and X_holdout["PLACEKEY"].isna().any():
            # CatBoost-only: its Pool rejects NaN in a cat_feature column,
            # unlike XGBoost/LightGBM which treat it as "missing" -- swap in
            # a sentinel category just for this predict call.
            holdout_X = X_holdout.copy()
            holdout_X["PLACEKEY"] = holdout_X["PLACEKEY"].cat.add_categories(["__unseen__"]).fillna("__unseen__")
        pred = np.expm1(model.predict(holdout_X))
    else:
        model = ordinal_pipeline(build_model(name))
        model.fit(train_df[FEATURE_COLUMNS], y_train)
        pred = np.expm1(model.predict(holdout_df[FEATURE_COLUMNS]))
    fitted[name] = model
    rows.append({"model": name, **score(y_holdout, pred)})

comparison = pd.DataFrame(rows).sort_values("MAE").reset_index(drop=True)
display(comparison)

winner_name = comparison.iloc[0]["model"]
print(f"\nBest model by MAE: {winner_name}")
```

## Step 5 — Refit the bake-off winner through 2024-12-31, generate one prediction per PLACEKEY per day for 2025

Rolls the winning model (or the lag-365 baseline itself, if nothing beat it in Step 4) forward one day at a time: each day's prediction feeds `visits_lag_1/7/28` for the next day, since there's no ground truth yet for a date that hasn't happened. `visits_lag_365` stays a *real* 2024 observation for the entire 2025 horizon, which is what keeps this from degrading the way a pure short-lag recursive rollout would — `classical_ts_daily.py`'s month-ahead bake-off already showed 31 days of that kind of compounding losing to a naive baseline, and 365 days on top of a short-lag model would likely be worse. If Step 4's winner didn't clearly beat the baseline, this cell is effectively just replaying the baseline forward.

PLACEKEYs with less than 365 days of history by the forecast origin are skipped (no real `visits_lag_365` to build from) rather than imputed.

```python
final_train = featurized.dropna(subset=NUMERIC_COLUMNS + [TARGET])
y_final = np.log1p(final_train[TARGET].to_numpy())

train_categories = None
if winner_name == "baseline_seasonal_lag365":
    final_model = None
elif winner_name in NATIVE_CATEGORICAL_MODELS:
    X_final, _ = native_categorical_frame(final_train, final_train)
    final_model = build_model(winner_name)
    final_model.fit(X_final, y_final)
    train_categories = X_final["PLACEKEY"].cat.categories
else:
    final_model = ordinal_pipeline(build_model(winner_name))
    final_model.fit(final_train[FEATURE_COLUMNS], y_final)


def recursive_placekey_forecast(history: pd.DataFrame, forecast_dates: pd.DatetimeIndex) -> pd.DataFrame:
    history = history.sort_values(["PLACEKEY", "LOCAL_DATE"])
    holidays = set(USFederalHolidayCalendar().holidays(start=forecast_dates.min(), end=forecast_dates.max()))

    series = {
        placekey: group[TARGET].astype(float).tolist()
        for placekey, group in history.groupby("PLACEKEY")
    }

    predictions = []
    for forecast_date in forecast_dates:
        rows = []
        placekeys = []
        for placekey, values in series.items():
            if len(values) < 365:
                continue  # no real visits_lag_365 yet for this store -- skip rather than impute
            rows.append({
                "PLACEKEY": placekey,
                "month": forecast_date.month,
                "year": forecast_date.year,
                "day_of_week": forecast_date.dayofweek,
                "day_of_month": forecast_date.day,
                "is_weekend": forecast_date.dayofweek >= 5,
                "is_federal_holiday": forecast_date in holidays,
                "visits_lag_1": values[-1],
                "visits_lag_7": values[-7],
                "visits_lag_28": values[-28],
                "visits_lag_365": values[-365],
                "visits_rolling_7": float(np.mean(values[-7:])),
                "visits_rolling_28": float(np.mean(values[-28:])),
            })
            placekeys.append(placekey)

        features = pd.DataFrame(rows)
        if winner_name == "baseline_seasonal_lag365":
            predicted = features["visits_lag_365"].to_numpy()
        else:
            if winner_name in NATIVE_CATEGORICAL_MODELS:
                features["PLACEKEY"] = pd.Categorical(features["PLACEKEY"], categories=train_categories)
                # CatBoost-only NaN-vs-unseen-category guard, same reasoning as Step 4.
                if winner_name == "CatBoost" and features["PLACEKEY"].isna().any():
                    features["PLACEKEY"] = features["PLACEKEY"].cat.add_categories(["__unseen__"]).fillna("__unseen__")
            predicted = np.clip(np.expm1(final_model.predict(features[FEATURE_COLUMNS])), 0, None)

        for placekey, value in zip(placekeys, predicted):
            value = float(value)
            series[placekey].append(value)
            predictions.append({
                "PLACEKEY": placekey,
                "LOCAL_DATE": forecast_date,
                f"predicted_{TARGET}": value,
            })
    return pd.DataFrame(predictions)


forecast_dates_2025 = pd.date_range("2025-01-01", "2025-12-31", freq="D")
history_through_origin = featurized[featurized["LOCAL_DATE"] <= FORECAST_ORIGIN]

# 365 sequential predict calls, one per PLACEKEY-day -- expect this cell to
# take a while on the full dataset.
predictions_2025 = recursive_placekey_forecast(history_through_origin, forecast_dates_2025)
print(f"Generated {len(predictions_2025):,} predictions across {predictions_2025['PLACEKEY'].nunique():,} PLACEKEYs")

# Absolute, matching JOINED_SPEND_FILE's style above -- a relative path
# here resolves against the KERNEL's cwd (this notebook's own directory
# when run via nbconvert), not the repo root, and silently wrote the
# first full run into visitor/visitor_data/... instead of visitor_data/...
OUTPUT_PATH = Path(
    "/Users/rusli/Documents/GitHub/Rice-To-Meet-You/"
    "visitor_data/output/visitor_predictions_2025.csv"
)
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)
predictions_2025.to_csv(OUTPUT_PATH, index=False)
print(f"Wrote -> {OUTPUT_PATH}")
```
